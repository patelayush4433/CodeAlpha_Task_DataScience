```python
import pandas as pd
```

```python
dataset = pd.read_csv('car data (1).csv')
```

```python
print("Columns:", dataset.columns.tolist())
print("\nDataset shape:", dataset.shape)
print("\nFirst few rows:")
print(dataset.head(3))
```

```python
dataset['Brand'] = dataset['Car_Name'].str.split().str[0]
```

```python
dataset["Brand"]
```

```python
features = ['Brand', 'Year', 'Selling_Price', 'Present_Price', 'Driven_kms', 
            'Fuel_Type', 'Selling_type', 'Transmission', 'Owner']
car_features = dataset[features].copy()
```

```python
summary = car_features.describe()
```

```python
summary
```

```python
brand_counts = car_features['Brand'].value_counts().head(10)
brand_counts
```

```python
print("\nBrand popularity (top 10):")
print(brand_counts)
```
