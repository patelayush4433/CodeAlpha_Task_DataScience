```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dataset = pd.read_csv("Unemployment in India.csv")
dataset.head(3)
```

```python
dataset.columns = dataset.columns.str.strip()
```

```python
dataset['Date'] = pd.to_datetime(dataset['Date'])
```

```python
print(dataset.isnull().sum())
```

```python
print(dataset['Estimated Unemployment Rate (%)'].describe())
```

```python
plt.figure(figsize=(12,6))

sns.lineplot(x='Date', y='Estimated Unemployment Rate (%)',data=dataset)

plt.title("Unemployment Rate During COVID-19")
plt.xticks(rotation=45)
plt.show()
```
