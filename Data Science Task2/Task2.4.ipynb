```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dataset = pd.read_csv("Unemployment in India.csv")
dataset.head(3)
```

```python
dataset[' Date'] = pd.to_datetime(dataset[' Date'])
```

```python
dataset = dataset.sort_values(by=' Date')
```

```python
plt.figure(figsize=(12,6))
plt.plot(dataset[' Date'], dataset[' Estimated Unemployment Rate (%)'], marker='o')
```

```python
plt.title("Unemployment Rate Trend in India")
plt.xlabel("Date")
plt.ylabel("Unemployment Rate (%)")
plt.xticks(rotation=45)

plt.grid(True)
plt.show()
```

```python
monthly_avg = dataset.groupby(dataset[' Date'].dt.month)[' Estimated Unemployment Rate (%)'].mean()

print("\nMonthly Average Unemployment Rate:")
print(monthly_avg)
```
