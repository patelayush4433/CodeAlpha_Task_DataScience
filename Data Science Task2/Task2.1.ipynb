```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dataset = pd.read_csv("Unemployment in India.csv")
dataset.head(3)
```

```python
print(dataset.info())
```

```python
dataset.isnull().sum()
```

```python
avg_rate = dataset[' Estimated Unemployment Rate (%)'].mean()
print("Average Unemployment Rate:", avg_rate)
```

```python
region_avg = dataset.groupby('Region')[' Estimated Unemployment Rate (%)'].mean()
print(region_avg)
```

```python
plt.figure(figsize=(12,6))
region_avg.sort_values().plot(kind='bar')
plt.title("Average Unemployment Rate by Region")
plt.xlabel("Region")
plt.ylabel("Unemployment Rate (%)")

plt.show()
```
